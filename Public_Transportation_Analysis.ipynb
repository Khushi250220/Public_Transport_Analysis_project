```python
data["departure_time"] = pd.to_datetime(data["departure_time"],
                                       format = "%H:%M").dt.time
```

```python
data['day_type'] = np.where(data["day_of_week"].isin(["Sunday","Saturday"]), 'Weekend',
    'Weekday')
```

```python
data['on_time_status'] = np.where(
    data['delay_minutes'] <= 5,
    'On Time',
    'Delayed'
)
```

```python
data['utilization_rate']
```

```python
data["utilization_category"] = pd.cut(
    data["utilization_rate"],
    bins= [0.0,0.50,0.75,1.00,np.inf],
    labels = ['Low','Moderate','High','High_Capacity']
)
```

# Start EDA 🔍

## Question 1

###  which routes have the highest passenger demand

```python
passanger_route = (
    data.groupby('route_name')['passengers'].sum().sort_values(ascending=False)
)
passanger_route.head(10)
```

## Question 2

### Which routes have the highest average delay?

```python
average_delay = (
    data.groupby('route_name')['delay_minutes'].mean().sort_values(ascending=False)
    
)
average_delay.head(10)
```

 ## Question 3

 ### What are the peak travel periods?

```python
peak_hour =(
    data.groupby('peak_hour')['passengers'].sum()
)
peak_hour
```

### Question 4

### Which routes are underutilized?

```python
route_underutilized =( data.groupby('route_name')['utilization_rate'].mean().sort_values(ascending=False))
route_underutilized.head(10)
```

### Question 5

### What is the cancellation rate?

```python
data_cancelled_rate = (data['trip_status'].value_counts('Cancelled')*100)

data_cancelled_rate
```

##  connect to SQL

```python
from getpass import getpass
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

MYSQL_USER = "root"
MYSQL_PASSWORD = getpass("Enter MySQL password: ")
MYSQL_HOST = "127.0.0.1"
MYSQL_PORT = 3306
DATABASE = "Public_transport"

# Connect to the MySQL server first, without selecting a database.
server_engine = create_engine(
    f"mysql+pymysql://{MYSQL_USER}:{MYSQL_PASSWORD}@{MYSQL_HOST}:{MYSQL_PORT}/",
    pool_pre_ping=True
)
```

```python
# Connect to the customer database
engine = create_engine(
    f"mysql+pymysql://{MYSQL_USER}:{MYSQL_PASSWORD}@{MYSQL_HOST}:{MYSQL_PORT}/{DATABASE}",
    pool_pre_ping=True
)
```

```python
table_name = "transport_table"

data.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)

print(f"Data successfully written to MySQL table: {table_name}")
```

```python
# Read the table back from MySQL
transport_table = pd.read_sql(
    text(f"SELECT * FROM `{table_name}`"),
    engine
)

Transport_table.head(15)
```
